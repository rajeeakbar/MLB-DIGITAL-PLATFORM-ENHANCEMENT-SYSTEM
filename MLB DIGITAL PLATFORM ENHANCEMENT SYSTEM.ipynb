```python

# Used to create multiple threads for report generation
import threading

# =========================================================
# PLAYER MANAGEMENT CLASS
# =========================================================

class Player:
    """
    Represents a baseball player.

    This class is responsible for storing player
    information and performance statistics.
    """

    def __init__(self, player_id, name, position, team):
        self.player_id = player_id
        self.name = name
        self.position = position
        self.team = team

        # Statistics start at zero
        self.runs = 0
        self.hits = 0

    def update_stats(self, runs, hits):
        """
        Updates player performance statistics.
        """

        self.runs += runs
        self.hits += hits

    def display_info(self):
        """
        Displays all player details.
        """

        return (
            f"ID: {self.player_id}, "
            f"Name: {self.name}, "
            f"Position: {self.position}, "
            f"Team: {self.team}, "
            f"Runs: {self.runs}, "
            f"Hits: {self.hits}"
        )
```

```python

# =========================================================
# TEAM MANAGEMENT CLASS
# =========================================================

class Team:
    """
    Represents an MLB team.

    A team contains many players.
    """

    def __init__(self, team_id, team_name):
        self.team_id = team_id
        self.team_name = team_name

        # List stores player objects
        self.players = []

    def add_player(self, player):
        """
        Adds a player to team roster.
        """
        self.players.append(player)

    def remove_player(self, player_id):
        """
        Removes player using player ID.
        """

        for player in self.players:
            if player.player_id == player_id:
                self.players.remove(player)
                return "Player removed."

        return "Player not found."

    def display_roster(self):
        """
        Displays all players in roster.
        """

        print(f"\nTeam: {self.team_name}")
        print("-" * 40)

        for player in self.players:
            print(player.display_info())
```

```python

# =========================================================
# MATCH SCHEDULE MANAGEMENT
# =========================================================

class Match:
    """
    Represents a scheduled baseball match.
    """

    def __init__(self,
                 match_id,
                 home_team,
                 away_team,
                 date,
                 venue):

        self.match_id = match_id
        self.home_team = home_team
        self.away_team = away_team
        self.date = date
        self.venue = venue

    def update_schedule(self, new_date):
        """
        Updates the match date.
        """

        self.date = new_date

    def display_match(self):
        """
        Displays match details.
        """

        return (
            f"Match ID: {self.match_id}\n"
            f"{self.home_team} vs {self.away_team}\n"
            f"Date: {self.date}\n"
            f"Venue: {self.venue}\n"
        )
```

```python

# =========================================================
# TICKET MANAGEMENT
# =========================================================

class Ticket:
    """
    Represents a baseball ticket.
    """

    def __init__(self,
                 ticket_id,
                 match,
                 price):

        self.ticket_id = ticket_id
        self.match = match
        self.price = price

        # Ticket starts in available state
        self.booked = False

    def book_ticket(self):
        """
        Books a ticket.
        """

        if self.booked:
            return "Ticket already booked."

        self.booked = True
        return "Ticket booked successfully."

    def cancel_ticket(self):
        """
        Cancels a booked ticket.
        """

        if not self.booked:
            return "Ticket was not booked."

        self.booked = False
        return "Ticket cancelled."

    def display_ticket(self):

        status = "Booked" if self.booked else "Available"

        return (
            f"Ticket ID: {self.ticket_id}\n"
            f"Price: ${self.price}\n"
            f"Status: {status}"
        )
```

```python

# =========================================================
# BOOKING MANAGEMENT
# =========================================================

class Booking:
    """
    Represents a customer booking.
    """

    def __init__(self,
                 booking_id,
                 customer_name,
                 ticket):

        self.booking_id = booking_id
        self.customer_name = customer_name
        self.ticket = ticket

    def display_booking(self):

        return (
            f"Booking ID: {self.booking_id}\n"
            f"Customer: {self.customer_name}\n"
            f"Ticket: {self.ticket.ticket_id}"
        )
```

```python

# =========================================================
# MLB BACKEND CLASS
# =========================================================

class MLBBackend:
    """
    Central management system.

    Stores all project data.
    """

    def __init__(self):

        self.players = []
        self.teams = []
        self.matches = []
        self.tickets = []
        self.bookings = []

    # ----------------------------------
    # PLAYER MANAGEMENT
    # ----------------------------------

    def add_player(self, player):
        self.players.append(player)

    # ----------------------------------
    # TEAM MANAGEMENT
    # ----------------------------------

    def add_team(self, team):
        self.teams.append(team)

    # ----------------------------------
    # MATCH MANAGEMENT
    # ----------------------------------

    def add_match(self, match):
        self.matches.append(match)

    # ----------------------------------
    # TICKET MANAGEMENT
    # ----------------------------------

    def add_ticket(self, ticket):
        self.tickets.append(ticket)

    # ----------------------------------
    # BOOKING MANAGEMENT
    # ----------------------------------

    def add_booking(self, booking):
        self.bookings.append(booking)

    # =====================================================
    # REPORTS
    # =====================================================

    def generate_player_report(self, player):
        """
        Thread task.

        Each thread generates a report
        for one player.
        """

        print("\nPLAYER REPORT")
        print("=" * 40)
        print(player.display_info())

    def generate_reports(self):
        """
        Generates reports simultaneously.

        This demonstrates multithreading.
        """

        threads = []

        for player in self.players:

            thread = threading.Thread(
                target=self.generate_player_report,
                args=(player,)
            )

            threads.append(thread)

            # Start executing thread
            thread.start()

        # Wait until all threads finish
        for thread in threads:
            thread.join()

        print("\nAll reports generated successfully.")


# =========================================================
# DRIVER PROGRAM
# =========================================================

if __name__ == "__main__":

    backend = MLBBackend()

    # =====================================================
    # CREATE PLAYERS
    # =====================================================

    player1 = Player(
        1,
        "Aaron Judge",
        "Outfielder",
        "Yankees"
    )

    player2 = Player(
        2,
        "Mookie Betts",
        "Shortstop",
        "Dodgers"
    )

    player1.update_stats(25, 40)
    player2.update_stats(30, 52)

    backend.add_player(player1)
    backend.add_player(player2)

    # =====================================================
    # CREATE TEAM
    # =====================================================

    yankees = Team(
        101,
        "New York Yankees"
    )

    yankees.add_player(player1)
    yankees.add_player(player2)

    backend.add_team(yankees)

    # =====================================================
    # CREATE MATCH
    # =====================================================

    match1 = Match(
        1001,
        "New York Yankees",
        "LA Dodgers",
        "15-12-2026",
        "Yankee Stadium"
    )

    backend.add_match(match1)

    # =====================================================
    # CREATE TICKET
    # =====================================================

    ticket1 = Ticket(
        5001,
        match1,
        120
    )

    print(ticket1.book_ticket())

    backend.add_ticket(ticket1)

    # =====================================================
    # CREATE BOOKING
    # =====================================================

    booking1 = Booking(
        8001,
        "Akbar Rajee",
        ticket1
    )

    backend.add_booking(booking1)

    # =====================================================
    # DISPLAY DATA
    # =====================================================

    print("\nMATCH DETAILS")
    print(match1.display_match())

    print("\nTICKET DETAILS")
    print(ticket1.display_ticket())

    print("\nBOOKING DETAILS")
    print(booking1.display_booking())

    print("\nTEAM ROSTER")
    yankees.display_roster()

    # =====================================================
    # MULTITHREADED REPORT GENERATION
    # =====================================================

    print("\nGENERATING REPORTS...")
    backend.generate_reports()
```
